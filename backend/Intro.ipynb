```python
# Load CSV files
dataset_1 = pd.read_csv("healthcare_dataset.csv")
dataset_2 = pd.read_csv("healthcare_noshows.csv")
dataset_3 = pd.read_csv("KaggleV2-May-2016.csv")
```

```python
# Display basic info
print(dataset_1.info())
print(dataset_2.info())
print(dataset_3.info())

# Show first few rows
print(dataset_1.head())
print(dataset_2.head())
print(dataset_3.head())
```

```python
# Check missing values
print(dataset_1.isnull().sum())
print(dataset_2.isnull().sum())
print(dataset_3.isnull().sum())

# Check for duplicates
print("Duplicates in dataset 1:", dataset_1.duplicated().sum())
print("Duplicates in dataset 2:", dataset_2.duplicated().sum())
print("Duplicates in dataset 3:", dataset_3.duplicated().sum())
```

```python
dataset_1_cleaned = dataset_1.drop_duplicates()
dataset_2_cleaned = dataset_2.drop_duplicates()
dataset_3_cleaned = dataset_3.drop_duplicates()
```

```python
# Convert date columns to datetime
dataset_2_cleaned["ScheduledDay"] = pd.to_datetime(dataset_2_cleaned["ScheduledDay"])
dataset_2_cleaned["AppointmentDay"] = pd.to_datetime(dataset_2_cleaned["AppointmentDay"])

dataset_3_cleaned["ScheduledDay"] = pd.to_datetime(dataset_3_cleaned["ScheduledDay"])
dataset_3_cleaned["AppointmentDay"] = pd.to_datetime(dataset_3_cleaned["AppointmentDay"])

# Convert PatientId to string
dataset_2_cleaned["PatientId"] = dataset_2_cleaned["PatientId"].astype(str)
dataset_3_cleaned["PatientId"] = dataset_3_cleaned["PatientId"].astype(str)

# Convert No-show column to binary
dataset_3_cleaned["No-show"] = dataset_3_cleaned["No-show"].map({"Yes": 1, "No": 0})
```

```python
# Normalize the values
def remove_outliers(df, numeric_cols):
    Q1 = df[numeric_cols].quantile(0.25)
    Q3 = df[numeric_cols].quantile(0.75)
    IQR = Q3 - Q1
    return df[~((df[numeric_cols] < (Q1 - 1.5 * IQR)) | (df[numeric_cols] > (Q3 + 1.5 * IQR))).any(axis=1)]

# Identify numerical columns
numeric_cols_1 = dataset_1_cleaned.select_dtypes(include=[np.number]).columns.tolist()
numeric_cols_2 = dataset_2_cleaned.select_dtypes(include=[np.number]).columns.tolist()
numeric_cols_3 = dataset_3_cleaned.select_dtypes(include=[np.number]).columns.tolist()

# Do NOT treat the target variable as an outlier
if "No-show" in numeric_cols_3:
    numeric_cols_3.remove("No-show")

# Apply outlier removal
dataset_1_no_outliers = remove_outliers(dataset_1_cleaned, numeric_cols_1)
dataset_2_no_outliers = remove_outliers(dataset_2_cleaned, numeric_cols_2)
dataset_3_no_outliers = remove_outliers(dataset_3_cleaned, numeric_cols_3)
```

```python
dataset_1_no_outliers.to_csv("healthcare_dataset_preprocessed.csv", index=False)
dataset_2_no_outliers.to_csv("healthcare_noshows_preprocessed.csv", index=False)
dataset_3_no_outliers.to_csv("KaggleV2-May-2016_preprocessed.csv", index=False)
print("Preprocessed datasets saved successfully!")
```
